```
import random

suits = ["hearts", "diamonds", "clubs", "spades"]
values = ["ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "jack", "queen", "king"]
deck = []
for suit in suits:
    for value in values:
        deck.append((value, suit))
random.shuffle(deck)
# 2 slicing
hand = deck[:5]
# 4. how many hearts
hand_suits = []
for value, suit in hand:
    hand_suits.append(suit)
hearts_count = hand_suits.count("hearts")
print(f"Hearts: {hearts_count}")
# 5. using in and count
picture_count = 0
for value, suit in hand:
    if value in ["jack", "queen", "king"]:
        picture_count += 1
print(f"Picture cards: {picture_count}")
# 6.if  true  flush or no flush
if hand_suits.count(hand_suits[0]) == 5:
    print("Flush!")
else:
    print("No flush")
```
